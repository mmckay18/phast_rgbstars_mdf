```python
start_time = time.time()
phat_gst_catalog_df = pd.read_csv('/Users/mmckay/phd_projects/analysis_routine/DATA/4GYRS_RGB_AGB_isochrones/phat_gst_optical_catalog.csv')
phast_gst_catalog_df = pd.read_csv('/Users/mmckay/phd_projects/analysis_routine/DATA/4GYRS_RGB_AGB_isochrones/phast_gst_optical_catalog.csv')
phast_gst_catalog_df.head()


shared_cols = phat_gst_catalog_df.columns.intersection(phast_gst_catalog_df.columns).tolist()
m31_gst_catalog_df = pd.concat([phat_gst_catalog_df[shared_cols], phast_gst_catalog_df[shared_cols]], ignore_index=True)
print(f"Number of rows in combined M31 GST catalog before deduplication: {len(m31_gst_catalog_df)}")
m31_gst_catalog_df.drop_duplicates(inplace=True)
print(f"Number of rows in combined M31 GST catalog after deduplication: {len(m31_gst_catalog_df)}")
m31_gst_catalog_df.reset_index(drop=True, inplace=True)
m31_gst_catalog_df.head()
m31_gst_catalog_df.to_csv(f'/Users/mmckay/phd_projects/analysis_routine/DATA/m31_photometry_gst_optical_catalog_{len(m31_gst_catalog_df)}.csv', index=False)
end_time = time.time()
print(f"Time taken to merge and save M31 GST catalog: {(end_time - start_time)/60:.8f} minutes")

```
